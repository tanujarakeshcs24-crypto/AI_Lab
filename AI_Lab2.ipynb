```python
def dls(state, goal, limit, path):
    if state == goal:
        return path

    if limit == 0:
        return None

    zero = state.index(0)
    r, c = divmod(zero, 3)

    for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            pos = nr * 3 + nc
            board = list(state)
            board[zero], board[pos] = board[pos], board[zero]
            new_state = tuple(board)

            if new_state not in path:
                result = dls(
                    new_state, goal, limit - 1,
                    path + [new_state]
                )

                if result is not None:
                    return result

    return None


def ids(start, goal):
    depth = 0

    while True:
        result = dls(start, goal, depth, [start])

        if result is not None:
            return result

        depth += 1


start = (5, 4, 6, 1, 8, 7, 3, 2, 0)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

result = ids(start, goal)

for state in result:
    print(state[:3])
    print(state[3:6])
    print(state[6:])
    print()
```

```python
def dfs(start, goal):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path + [state]

        if tuple(state) in visited:
            continue

        visited.add(tuple(state))

        zero = state.index(0)
        row, col = zero // 3, zero % 3

        moves = [
            (-1, 0),  # Up
            (1, 0),   # Down
            (0, -1),  # Left
            (0, 1)    # Right
        ]

        for dr, dc in moves:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                new_pos = r * 3 + c
                new_state = state.copy()

                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )

                if tuple(new_state) not in visited:
                    stack.append((new_state, path + [state]))

    return None


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

result = dfs(start, goal)

if result:
    print("Solution found:")
    for state in result:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
else:
    print("No solution found")
```
