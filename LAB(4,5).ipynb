```python

N = 4


def calculate_cost(state):
    """Count the number of attacking queen pairs."""
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            if (state[i] == state[j] or
                    abs(state[i] - state[j]) == abs(i - j)):
                cost += 1

    return cost


def generate_neighbors(state):
    """Generate all states by moving one queen."""
    neighbors = []

    for col in range(N):
        for row in range(1, N + 1):
            if row != state[col]:
                neighbor = state.copy()
                neighbor[col] = row
                neighbors.append(neighbor)

    return neighbors


def hill_climbing(initial_state):
    current = initial_state.copy()
    iteration = 0

    while True:
        current_cost = calculate_cost(current)

        print(f"\nIteration {iteration}")
        print("Current state:", current)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("\nGoal configuration reached!")
            print("Final state:", current)
            print_board(current)
            return current

        neighbors = generate_neighbors(current)

        for neighbor in neighbors:
            print(neighbor, "-> Cost:",
                  calculate_cost(neighbor))

        best_neighbor = min(
            neighbors, key=calculate_cost
        )
        best_cost = calculate_cost(best_neighbor)

        print("Best neighbor:", best_neighbor)
        print("Best neighbor cost:", best_cost)

        if best_cost >= current_cost:
            print("Local minimum reached.")
            print("No better neighbor exists.")
            return None

        current = best_neighbor
        iteration += 1


def print_board(state):
    """Display the chessboard."""
    print("\nChessboard:")

    for row in range(1, N + 1):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

initial_state = [4, 2, 3, 1]

hill_climbing(initial_state)
```

```python

N = 4

def cost(state):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                count += 1
    return count

def print_board(state):
    for row in range(1, N + 1):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

def neighbors(state):
    result = []
    for col in range(N):
        for row in range(1, N + 1):
            if row != state[col]:
                new_state = state.copy()
                new_state[col] = row
                result.append(new_state)
    return result

def hill_climbing(state):
    current = state.copy()
    step = 0

    while True:
        current_cost = cost(current)

        print("\nStep:", step)
        print("State:", tuple(current))
        print("Cost:", current_cost)
        print_board(current)

        if current_cost == 0:
            print("Goal state reached!")
            break

        next_states = neighbors(current)
        best = min(next_states, key=cost)
        best_cost = cost(best)

        print("\nNeighbor boards:")
        for s in next_states:
            print("\nState:", tuple(s), "Cost:", cost(s))
            print_board(s)

        if best_cost >= current_cost:
            print("\nLocal minimum reached. No better neighbor.")
            break

        current = best
        step += 1

initial_state = [4, 2, 3, 1]
hill_climbing(initial_state)
```

```python

import random
import math

N = 8

def calculate_cost(state):
    """Count pairs of queens attacking each other."""
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            # Queens attack if they share a diagonal.
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def get_neighbor(state):
    """Move one queen to a different row in its column."""
    neighbor = state.copy()
    col = random.randrange(N)

    rows = [r for r in range(N) if r != state[col]]
    neighbor[col] = random.choice(rows)

    return neighbor


def simulated_annealing():
    temperature = 100.0
    cooling_rate = 0.995
    min_temperature = 0.001
    max_steps = 200000

    current = [random.randrange(N) for _ in range(N)]
    current_cost = calculate_cost(current)

    best = current.copy()
    best_cost = current_cost

    for step in range(1, max_steps + 1):
        if current_cost == 0:
            return current, step

        if temperature < min_temperature:
            break

        next_state = get_neighbor(current)
        next_cost = calculate_cost(next_state)

        delta_e = current_cost - next_cost

        # Accept a better state or sometimes a worse state.
        if delta_e > 0 or random.random() < math.exp(
            delta_e / temperature
        ):
            current = next_state
            current_cost = next_cost

        if current_cost < best_cost:
            best = current.copy()
            best_cost = current_cost

        temperature *= cooling_rate

    if best_cost == 0:
        return best, step

    return None, step


def display_board(state):
    if state is None:
        print("Solution not found. Try again.")
        return

    print("\nChessboard solution:")
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

    print("\nQueen positions (row for each column):", state)
    print("Number of attacking pairs:", calculate_cost(state))


if __name__ == "__main__":
    random.seed()  # Use a different random sequence on each run.
    solution, steps = simulated_annealing()

    print("Simulated Annealing - 8 Queens")
    print("Steps:", steps)
    display_board(solution)
```
