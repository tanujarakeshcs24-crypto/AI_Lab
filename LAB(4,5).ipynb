```python

N = 4

def cost(state):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                count += 1
    return count


def print_board(state):
    print("+---+---+---+---+")
    for row in range(1, N + 1):
        print("|", end=" ")
        for col in range(N):
            if state[col] == row:
                print("Q |", end=" ")
            else:
                print(". |", end=" ")
        print()
        print("+---+---+---+---+")


def neighbors(state):
    result = []
    for col in range(N):
        for row in range(1, N + 1):
            if row != state[col]:
                new_state = state.copy()
                new_state[col] = row
                result.append(new_state)
    return result


def hill_climbing(state):
    current = state.copy()
    step = 0

    while True:
        current_cost = cost(current)

        print(f"\nStep: {step}")
        print(f"State: {tuple(current)}")
        print(f"Cost: {current_cost}")
        print_board(current)

        if current_cost == 0:
            print("Goal state reached!")
            break

        next_states = neighbors(current)
        best = min(next_states, key=cost)
        best_cost = cost(best)

        print("\nNeighbor Boards:")

        for s in next_states:
            print(f"\nState: {tuple(s)} | Cost: {cost(s)}")
            print_board(s)

        if best_cost >= current_cost:
            print("\nLocal minimum reached. No better neighbor.")
            break

        current = best
        step += 1


initial_state = [4, 2, 3, 1]
hill_climbing(initial_state)
```

```python
import random
import math

N = 8

# Calculate attacking pairs
def calculate_cost(board):
    cost = 0
    for i in range(N):
        for j in range(i + 1, N):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                cost += 1
    return cost


# Display chessboard
def display_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


def simulated_annealing():
    current = [random.randrange(N) for _ in range(N)]
    current_cost = calculate_cost(current)

    best = current.copy()
    best_cost = current_cost

    temperature = 100.0
    cooling_rate = 0.99
    min_temperature = 0.001
    iteration = 0

    print("Initial Board:")
    display_board(current)
    print("Initial Cost:", current_cost)

    while temperature > min_temperature:
        if current_cost == 0:
            break

        # Generate a neighbor by moving one queen
        neighbor = current.copy()
        col = random.randrange(N)
        row = random.randrange(N - 1)

        # Ensure the new row differs from the current row
        if row >= neighbor[col]:
            row += 1

        neighbor[col] = row
        neighbor_cost = calculate_cost(neighbor)

        delta = neighbor_cost - current_cost

        # Accept better moves or some worse moves
        if (delta < 0 or
            random.random() < math.exp(-delta / temperature)):
            current = neighbor
            current_cost = neighbor_cost

        # Track the best state found
        if current_cost < best_cost:
            best = current.copy()
            best_cost = current_cost

        iteration += 1

        print("\n" + "=" * 35)
        print("Iteration:", iteration)
        print("Temperature:", round(temperature, 4))
        print("Current Cost:", current_cost)
        print("Neighbor Cost:", neighbor_cost)
        print("Best Cost:", best_cost)

        # Print the current chessboard after every iteration
        print("\nCurrent Board:")
        display_board(current)

        # Cool down
        temperature *= cooling_rate

    print("\nFinal Best State:", best)
    print("Final Minimum Cost:", best_cost)
    print("Total Iterations:", iteration)

    print("\nFinal Best Board:")
    display_board(best)

    if best_cost == 0:
        print("Solution found!")
    else:
        print("No solution found in this run. Run again.")


simulated_annealing()
```
