```python

import heapq

start = (2, 8, 3, 1, 6, 4, 0, 7, 5)
goal = (1, 2, 3, 8, 0, 4, 7, 6, 5)

def misplaced(state):
    return sum(1 for i in range(9) if state[i] != 0 and state[i] != goal[i])

def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    moves = [-3, 3, -1, 1]

    for move in moves:
        new_pos = blank + move

        if 0 <= new_pos < 9:
            if move == -1 and blank % 3 == 0:
                continue
            if move == 1 and blank % 3 == 2:
                continue

            new_state = list(state)
            new_state[blank], new_state[new_pos] = new_state[new_pos], new_state[blank]
            neighbors.append(tuple(new_state))

    return neighbors

def astar():
    open_list = [(misplaced(start), 0, start)]
    parent = {start: None}
    cost = {start: 0}
    visited = set()

    while open_list:
        f, g, state = heapq.heappop(open_list)

        if state in visited:
            continue

        visited.add(state)

        if state == goal:
            path = []
            while state is not None:
                path.append(state)
                state = parent[state]

            path.reverse()

            print("Solution found!")
            print("Number of moves:", len(path) - 1)

            for step, state in enumerate(path):
                print("\nStep", step)
                for i in range(0, 9, 3):
                    print(state[i:i + 3])

            return

        for neighbor in get_neighbors(state):
            new_g = g + 1

            if neighbor not in cost or new_g < cost[neighbor]:
                cost[neighbor] = new_g
                parent[neighbor] = state
                new_f = new_g + misplaced(neighbor)
                heapq.heappush(open_list, (new_f, new_g, neighbor))

    print("No solution found")

astar()
```

```python

import heapq

start = (2, 8, 3, 1, 6, 4, 0, 7, 5)
goal = (1, 2, 3, 8, 0, 4, 7, 6, 5)

def manhattan(state):
    distance = 0

    for i in range(9):
        tile = state[i]

        if tile != 0:
            goal_pos = goal.index(tile)
            row1, col1 = divmod(i, 3)
            row2, col2 = divmod(goal_pos, 3)

            distance += abs(row1 - row2) + abs(col1 - col2)

    return distance

def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    moves = [-3, 3, -1, 1]

    for move in moves:
        new_pos = blank + move

        if 0 <= new_pos < 9:
            if move == -1 and blank % 3 == 0:
                continue
            if move == 1 and blank % 3 == 2:
                continue

            new_state = list(state)
            new_state[blank], new_state[new_pos] = new_state[new_pos], new_state[blank]
            neighbors.append(tuple(new_state))

    return neighbors

def astar():
    open_list = [(manhattan(start), 0, start)]
    parent = {start: None}
    cost = {start: 0}
    visited = set()

    while open_list:
        f, g, state = heapq.heappop(open_list)

        if state in visited:
            continue

        visited.add(state)

        if state == goal:
            path = []

            while state is not None:
                path.append(state)
                state = parent[state]

            path.reverse()

            print("Solution found!")
            print("Number of moves:", len(path) - 1)

            for step, state in enumerate(path):
                print("\nStep", step)

                for i in range(0, 9, 3):
                    print(state[i:i + 3])

            return

        for neighbor in get_neighbors(state):
            new_g = g + 1

            if neighbor not in cost or new_g < cost[neighbor]:
                cost[neighbor] = new_g
                parent[neighbor] = state
                new_f = new_g + manhattan(neighbor)
                heapq.heappush(open_list, (new_f, new_g, neighbor))

    print("No solution found")

astar()
```
